```python
print("Please select an option from the menu:")
print("1- If there are any problems or unclear things in the report, press 1")
print("2- If the report is late, press 2")
print("3- If you want to book an appointment, press 3")
print("4- If you want any other transaction, press 4")

choice = input("Enter your choice: ")

if choice == '1':
        print(" The doctor will call you when it is possible.")
else:
        if choice == '2':
            file_number = input(" Please enter the number of your file: ")
            print(f"File number {file_number} entered.")
            print("The report will be sent as soon as possible")
        else:
            if choice == '3':
                date = input("Please enter the date you want to book: ")
                print(f"Date {date} entered.")
                time = input("Please enter the time you want to book: ")
                print(f"Time {time} entered.")
                print("Your appointment has been booked.")
            else:
                if choice == '4':
                    print("You selected option 4: Please visit us.")
                else:
                    print("Invalid choice. Please select a valid option from the menu.")
```
